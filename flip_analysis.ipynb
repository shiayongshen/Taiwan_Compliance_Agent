```python
import pandas as pd
from pathlib import Path

def filter_sat_observations(excel_path="outputs_RQ3_llm_correction/llm_flips_analysis.xlsx"):
    """
    從 Excel 的 Details sheet 中篩選出 ground_truth_result 和 llm_validation_result 都是 SAT 的 observations
    
    參數:
        excel_path: Excel 文件路徑
    
    返回:
        包含篩選結果的 DataFrame
    """
    try:
        # 讀取 Details sheet
        df = pd.read_excel(excel_path, sheet_name='Details')
        
        # 篩選條件：兩個欄位都是 SAT
# 篩選條件：兩個欄位都是 SAT，且排除 case_324
        filtered_df = df[
            (df['ground_truth_result'].str.strip().str.upper() == 'SAT') & 
            (df['llm_validation_result'].str.strip().str.upper() == 'SAT') &
            (df['case_id'] != 'case_324')
        ]
        
        print(f"總共找到 {len(filtered_df)} 個 observation，其中兩個結果都是 SAT")
        print(f"\n篩選結果：")
        print(filtered_df)
        
        # 存成 CSV
        output_path = Path("sat_observations.csv")
        filtered_df.to_csv(output_path, index=False, encoding='utf-8-sig')
        print(f"\n結果已保存到: {output_path}")

        
        return filtered_df
        
    except Exception as e:
        print(f"錯誤: {e}")
        return None

if __name__ == "__main__":
    result = filter_sat_observations()
```

```python
import pandas as pd
from pathlib import Path
def compare_flips():
    """
    比較 smt_flips 和 llm_flips 的平均次數，並計算 flip rate
    """
    try:
        # 讀取 CSV
        df = pd.read_csv("sat_observations.csv")
        
        # 計算平均值
        avg_llm_flips = df['llm_flips'].mean()
        avg_smt_flips = df['smt_flips'].mean()
        
        # 計算 flip rate
        avg_llm_flip_rate = (df['llm_flips'] / df['total_facts']).mean()
        avg_smt_flip_rate = (df['smt_flips'] / df['total_facts']).mean()
        
        # 計算總數
        total_rows = len(df)
        
        print(f"總共有 {total_rows} 筆 observations")
        print(f"\n平均 llm_flips: {avg_llm_flips:.2f}")
        print(f"平均 smt_flips: {avg_smt_flips:.2f}")
        print(f"\n差異: {abs(avg_llm_flips - avg_smt_flips):.2f}")
        
        # 詳細統計
        print(f"\n=== llm_flips 統計 ===")
        print(f"最小值: {df['llm_flips'].min()}")
        print(f"最大值: {df['llm_flips'].max()}")
        print(f"中位數: {df['llm_flips'].median()}")
        print(f"標準差: {df['llm_flips'].std():.2f}")
        print(f"平均 flip rate: {avg_llm_flip_rate:.4f} ({avg_llm_flip_rate*100:.2f}%)")
        
        print(f"\n=== smt_flips 統計 ===")
        print(f"最小值: {df['smt_flips'].min()}")
        print(f"最大值: {df['smt_flips'].max()}")
        print(f"中位數: {df['smt_flips'].median()}")
        print(f"標準差: {df['smt_flips'].std():.2f}")
        print(f"平均 flip rate: {avg_smt_flip_rate:.4f} ({avg_smt_flip_rate*100:.2f}%)")
        
        # 比較分布
        print(f"\n=== 比較結果 ===")
        if avg_llm_flips > avg_smt_flips:
            print(f"LLM 的平均 flips 數比 SMT 多 {avg_llm_flips - avg_smt_flips:.2f}")
        else:
            print(f"SMT 的平均 flips 數比 LLM 多 {avg_smt_flips - avg_llm_flips:.2f}")
        
        if avg_llm_flip_rate > avg_smt_flip_rate:
            print(f"LLM 的平均 flip rate 比 SMT 高 {(avg_llm_flip_rate - avg_smt_flip_rate)*100:.2f}%")
        else:
            print(f"SMT 的平均 flip rate 比 LLM 高 {(avg_smt_flip_rate - avg_llm_flip_rate)*100:.2f}%")
        
        return df
        
    except Exception as e:
        print(f"錯誤: {e}")
        return None

if __name__ == "__main__":
    result = compare_flips()
```

```python
import pandas as pd
from pathlib import Path

def filter_sat_observations_only_SMT(excel_path="outputs_RQ3_llm_correction/llm_flips_analysis.xlsx"):
    """
    從 Excel 的 Details sheet 中篩選出 ground_truth_result 是 SAT 的 observations
    
    參數:
        excel_path: Excel 文件路徑
    
    返回:
        包含篩選結果的 DataFrame
    """
    try:
        # 讀取 Details sheet
        df = pd.read_excel(excel_path, sheet_name='Details')
        
        # 篩選條件：ground_truth_result欄位是 SAT
        filtered_df = df[
            (df['ground_truth_result'].str.strip().str.upper() == 'SAT') & 
            (df['case_id'] != 'case_324')
        ]
        
        print(f"總共找到 {len(filtered_df)} 個 observation，結果是 SAT")
        print(f"\n篩選結果：")
        print(filtered_df)
        
        # 存成 CSV
        output_path = Path("sat_observations_onlySMT.csv")
        filtered_df.to_csv(output_path, index=False, encoding='utf-8-sig')
        print(f"\n結果已保存到: {output_path}")

        
        return filtered_df
        
    except Exception as e:
        print(f"錯誤: {e}")
        return None

if __name__ == "__main__":
    result = filter_sat_observations_only_SMT()
```

```python
import pandas as pd
from pathlib import Path
def compare_flips_between_has_case_fact_or_not():
    """
    比較有hard constraint 或沒有的平均次數，並計算 flip rate
    """
    try:
        # 讀取 CSV
        df = pd.read_csv("sat_observations_onlySMT.csv")
        with_hard_df = df[(df['num_hard_constraint_facts']>0)]
        no_hard_df = df[(df['num_hard_constraint_facts']==0)]
        
        # 計算平均值
        avg_with_hard_flips = with_hard_df['smt_flips'].mean()
        avg_no_hard_flips = no_hard_df['smt_flips'].mean()
        avg_all_flip = df['smt_flips'].mean()
        # 計算 flip rate
        avg_with_hard_flip_rate = (with_hard_df['smt_flips'] / (with_hard_df['total_facts']-with_hard_df['num_hard_constraint_facts'])).mean()
        avg_no_hard_flip_rate = (no_hard_df['smt_flips'] / no_hard_df['total_facts']).mean()
        avg_all_flip_rate = (df['smt_flips'] / (df['total_facts']-df['num_hard_constraint_facts'])).mean()
        # 計算總數
        total_rows = len(df)
        
        print(f"總共有 {total_rows} 筆 observations")
        print(f"\n平均 with_hard_flips: {avg_with_hard_flips:.2f}")
        print(f"平均 no_hard_flips: {avg_no_hard_flips:.2f}")
        print(f"\n平均 all flip: {avg_all_flip:.2f}")
        print(f"\n差異: {abs(avg_with_hard_flips - avg_no_hard_flips):.2f}")
        
        # 詳細統計
        print(f"\n=== with hard 統計 ===")
        print(f"最小值: {with_hard_df['smt_flips'].min()}")
        print(f"最大值: {with_hard_df['smt_flips'].max()}")
        print(f"中位數: {with_hard_df['smt_flips'].median()}")
        print(f"標準差: {with_hard_df['smt_flips'].std():.2f}")
        print(f"平均 flip rate: {avg_with_hard_flip_rate:.4f} ({avg_with_hard_flip_rate*100:.2f}%)")
        
        print(f"\n=== without 統計 ===")
        print(f"最小值: {no_hard_df['smt_flips'].min()}")
        print(f"最大值: {no_hard_df['smt_flips'].max()}")
        print(f"中位數: {no_hard_df['smt_flips'].median()}")
        print(f"標準差: {no_hard_df['smt_flips'].std():.2f}")
        print(f"平均 flip rate: {avg_no_hard_flip_rate:.4f} ({avg_no_hard_flip_rate*100:.2f}%)")
        
        print(f"\n=== all 統計 ===")
        print(f"最小值: {df['smt_flips'].min()}")
        print(f"最大值: {df['smt_flips'].max()}")
        print(f"中位數: {df['smt_flips'].median()}")
        print(f"標準差: {df['smt_flips'].std():.2f}")
        print(f"平均 flip rate: {avg_all_flip_rate:.4f} ({avg_all_flip_rate*100:.2f}%)")
        
        
        return df
        
    except Exception as e:
        print(f"錯誤: {e}")
        return None

if __name__ == "__main__":
    result = compare_flips_between_has_case_fact_or_not()
```
